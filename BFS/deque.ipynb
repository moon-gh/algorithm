1. 준비물 챙기기 : 지도 크기(w,h), 방문 기록장(visited) 백지로 만들기
2. 첫 손님 입장 : 대기열(deque)에 시작점 넣고, 방문 도장 찍기
3. 방향 지시서 : 상하좌우 4방향 찌를 준비 (directions)
4. 본격 탐험 (while루프) <br>
: 맨 앞사람 꺼내서 도착지인지 확인,<br>
: 도착지가 아니면 4방향 찔럽기 <br>
(안전 검사 통과하면 대기열에 추가하고 도장 찍기)
5. 실패 선언

```python
from collections import deque

def solution(maps):
    h = len(maps)
    w = len(maps[0])
    
    visited = [[False] * w for _ in range(h)]
    
    queue = deque([(0, 0, 1)])
    visited[0][0] = True
    
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    
    while queue:
        r, c, dist = queue.popleft()
        
        if r == h-1 and c == w-1:
            return dist
        ste
        for dr, dc in directions:
            nr = r + dr
            nc = c + dc
            
            # 안전 검사
            if 0<=nr<h and 0<=nc<w and maps[nr][nc] == 1:
                if not visited[nr][nc]:
                    visited[nr][nc] = True
                    queue.append((nr, nc, dist+1))
                    
    return -1
```
